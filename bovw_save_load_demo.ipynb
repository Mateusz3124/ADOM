# Bag of Visual Words save and load demo

This notebook trains a BoVW model, saves it to disk, loads it back, and checks that the reloaded model behaves the same as the original model.

```python
from pathlib import Path

from bovw import BagOfVisualWordsClassifier, BoVWConfig
```

```python
dataset_root = Path('caltech101_reduced')
model_path = Path('artifacts') / 'bovw_model.pkl'

config = BoVWConfig(
    feature_extractor='SIFT',
    vocabulary_size=128,
    max_descriptors_per_image=250,
    max_vocabulary_descriptors=20000,
    random_state=42,
)
model = BagOfVisualWordsClassifier(config)
model.fit(dataset_root / 'train')
saved_path = model.save(model_path)
saved_path
```

```python
loaded_model = BagOfVisualWordsClassifier.load(saved_path)
original_metrics = model.evaluate(dataset_root / 'test')
loaded_metrics = loaded_model.evaluate(dataset_root / 'test')

print(f'Original accuracy: {original_metrics["accuracy"]:.4f}')
print(f'Loaded accuracy:   {loaded_metrics["accuracy"]:.4f}')
print('Predictions match:', (original_metrics['y_pred'] == loaded_metrics['y_pred']).all())
```

```python
sample_image = next((dataset_root / 'test').glob('*/*'))
print('Sample image:', sample_image)
print('Original prediction:', model.classify_image(sample_image))
print('Loaded prediction:', loaded_model.classify_image(sample_image))
```
